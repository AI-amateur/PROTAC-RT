```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('R2.csv')
type=df['algorithm']
PActivity=df['R2']
```

```python
import matplotlib as mpl
mpl.rcParams['font.family'] = 'Times'
plt.style.use('ggplot')

plt.xticks(
    fontsize=13,  
    color='black',    
)
plt.yticks(
    fontsize=13,  
    color='black',     
)



sns.violinplot(
    x=type,
    y=df.R2,    
    scale='width',
    bw=0.1,
    width=0.8,
    palette='Set3'

)
plt.ylabel(r'$R^2$', fontsize=14, color='black')  # 使用 LaTeX 表达式设置 R²
plt.xlabel('Algorithm', fontsize=12, color='black')
plt.savefig('R2.tif', dpi=300)
plt.show()
```
